### Reshaping Methods

#### reshape()

#### ravel()
Converts a multi-dimensional array into a 1D array
but ravel() returns a view whenever possible
so changing array can affect outputs

```python
a = np.arange(10,50,10).reshape(2,2)
print("array a:",a)
b = a.ravel()
print("array b:",b)
```

```python
b[0] = 100
# so now 
print(b)
print(a)
```

Also create 1d
but flatten() creates a copy

```python
a = np.arange(10,50,10).reshape(2,2)
print(a)
b = a.flatten()
print(b)
```

#### Copy vs View 
copy :
creates independent memory
modifying copy does not affect the original

view :
shares memory with original array 
modifying view affects original

```python
a = np.array([1,2,3])
view_array = a.view()
copy_array = a.copy()

print("orginal:",a)
print("original view:",view_array)
print("original copy:",copy_array)
print("after modifying")
a[0] = 100
print("original:",a)
print("view:",view_array)
print("copy:",copy_array)
```

# Handling NaN values

* np.nanmean()
* np.nanstd()
* np.nansum()
These functions ignore NaN values during computation

```python
a = np.array([10,20,np.nan,40])
print("Array:",a)
print("Mean (normal):",np.mean)
print("Mean ignoring NaN:",np.nanmean(a))

print("Std normal:",np.std(a))
print("std ignoring NaN:",np.nanstd(a))

print("Sum normal):",np.sum(a))
print("Sum ignoring NaN:",np.nansum(a))

#if in data set there are nan value and we use normal method result may be just NaN
```

# Data Cleaning

```python
prices = np.array([100,200,150,300,np.nan,250])
prices
```

```python
# Detect missing values
a = np.isnan(prices)
# Calculating avg price ignoring NaN
b =  np.nanmean(prices)
# Replace NaN with avg 
prices = np.where(a,b,prices)
print(prices)
```

## Identify Best Selling Product
use np.argmax()
This returns the index of the maximum value


```python
print("Best selling product index",np.argmax(prices))
print("Revenue generated:",prices[np.argmax(prices)])
```
