14.create array from 1 to 9 and 
* ravel()
* flatten()
Explain the difference

```python
a = np.arange(1,10)
print(a)
b = a.ravel()
print(b)
b[0] = 10
print("b:",b)
print("a:",a) 
```

```python
a = np.arange(1,10).reshape(3,3)
print(a)
b = a.flatten()
print(b)
b[0] = 10
print("b:",b)
print("a:",a)
```

```python
#difference is that ravel affect original
#while flatten dont affect original
```
