Histogram 
* A histogram is a univariate plot which helps us understand the distribution of continuous numerical variable 
* It consists of bars,where each bar represents the frequency (or count) of data points falling within a specific range or interval knowns as a bin

```python
import pandas as pd 
#import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
data = pd.read_csv('Automobile.csv')
```

```python
df = data 
sns.histplot(data=df,x='price');
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(6,4))
plt.title('Histogram: Price')
plt.xlim(5000, 50000)
# xlim min value 5000 and max 50000 in x-axis of
plt.ylim(0, 70)
plt.xlabel('Price of Cars')
plt.ylabel('Frequency')

sns.histplot(data=df, x='price', color='orange')

plt.show()
```

We can specify the number of intervals (or groups or bins) to create by setting the bins parameter

```python
sns.histplot(data=df,x='price',bins=5)
```

```python
sns.histplot(data=df,x='price',bins=5)
```

if we want to specify the width of the interval (or groups or bins) we can use binwidth parameter

```python
sns.histplot(data=df,x='price',binwidth=200)
```

python automatically define bins and it's mostly correct 

How to find the optimal number of bins: Rule of thumb

Binwidth = 2 *IQR 
        ----------------------
            cube root of n 
n = number of rows 
bins = Range
        ----------------------
                Binwidth
range = max-min 


### Histogram with Density Curve

* Kernel Density Estimation(Kde) visualizes the distribution of data over a continuous internal 
* the convertional scale for kde is : total frequency of each bin x probability 

```python
sns.histplot(data=df,x='price',kde=True)
```

```python
sns.histplot(data=df,x='price',bins=700,kde=True)
```

Let's check out the histogram for a few more attribute in the data

```python
sns.histplot(data=df,x='curb_weight',kde=True)
```

```python
sns.histplot(data=df,x='horsepower',kde=True)
```

Histogram are Intuitive but it is hardly a good choice when we want to compare the distributions of several groups

```python
sns.histplot(data=df,x='price',hue='body_style',kde=True)
```
