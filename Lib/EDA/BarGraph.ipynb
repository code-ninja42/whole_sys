```python
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
df = pd.read_csv('Automobile.csv')
```

Bar Graph 
* bar graph is generally used to show the counts of observation in each bin of categorical variable using bars
* we can use the countplot() function of seaborn to plot a bar graph

```python
sns.countplot(data=df,x='body_style');
```

```python
df['body_style'].value_counts()
```

```python
sns.countplot(data=df,x='body_style',hue='fuel_type');
```

```python
df['fuel_type'].value_counts()
```

```python
sns.countplot(data=df,x='make');
```

To avoid this messy and congested 
increase the size of plot using plt.figure

```python
plt.figure(figsize=(20,7))
sns.countplot(data=df,x='make');
plt.xticks(rotat=90);#xlabel up and down wisse
```

```python
plt.figure(figsize=(10,7))
sns.countplot(data=df,x='engine_type',hue='fuel_type');
```
