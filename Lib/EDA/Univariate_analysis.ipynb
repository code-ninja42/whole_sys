* Univarite analysis invovles examining and describing the distribution,central tendency and dispersion of a single variable in a dataset.

```python
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 
df = pd.read_csv('Melbourne_Housing.csv')
```

Observations as Distance

```python
df.head()
sns.histplot(data=df,x='Distance',stat='density')
plt.show()
sns.boxplot(data=df,x='Distance')
plt.show()
```

* The distribution is skewed towards right
* There are many outliers presents in this column
* Values above 25 Kms are being represented as outliers in the boxplot, indicating there the are many properties that are atleast 25 Kms away from CBD.

Observation on Landsize

```python
sns.displot(data=df,x='Landsize',kind='kde')
plt.show()
sns.boxplot(data=df,x='Landsize')
plt.show()
df['Landsize'].dtype
```

```python
# converting Landsize to sq. kilometers from sq. meters
#sns.distplot(data=df,x=df['Landsize']/10000,kind='kde')
sns.boxplot(data=df,x=df['Landsize']/100000)
```

Observations on BuildingArea

```python
import numpy as np
print(df['BuildingArea'].dtype)
df['BuildingArea'].astype(float)
```

```python
# df['BuildingArea'].value_counts()
# df['BuildingArea'].unique()
df['BuildingArea'] = df['BuildingArea'].replace('missing',np.nan)
#pd.numeric errors = 'coerce' converts anything that is invalid like missing NA abc into NaN
```

```python
df['BuildingArea']= df['BuildingArea'].astype(float)
```

```python
df['BuildingArea'] = df['BuildingArea'].replace(np.inf,np.nan
)
```

```python
sns.histplot(data=df, x='BuildingArea', kde=True)
plt.show()
```
