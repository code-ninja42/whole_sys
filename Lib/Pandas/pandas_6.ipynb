```python
import pandas as pd
import numpy as np

# Customer dataset
customers = pd.DataFrame({
    "Customer_ID": [1,2,3,4,5,6],
    "Name": ["Aman Sharma","Priya Patel","Rahul Verma","Sneha Shah","Karan Mehta","Riya Jain"],
    "Region": ["North","South","East","West","North",None]
})

# Orders dataset
orders = pd.DataFrame({
    "Customer_ID": [1,1,2,3,4,4,5,6,6],
    "Product": ["Laptop","Mouse","Laptop","Tablet","Laptop","Tablet","Mouse","Laptop","Mouse"],
    "Category": ["Electronics","electronics","Electronics","electronics","ELECTRONICS","electronics","electronics","Electronics","electronics"],
    "Amount": [50000,2000,55000,30000,60000,25000,1500,70000,np.nan],
    "Order_Date": ["2024-01-01","2024-01-05","2024-02-01","2024-02-10","2024-03-01","2024-03-05","2024-03-10","2024-04-01","2024-04-05"]
})

customers, orders
```

```python
# 1. Data Cleaning
customers.columns = customers.columns.str.strip()
orders.columns = orders.columns.str.strip()

# 2. Missing Value Handling
customers['Region'].fillna('Unknown', inplace=True)

# Fill missing Amount with mean
orders['Amount'].fillna(orders['Amount'].mean(), inplace=True)

# 3. String Normalization
orders['Category'] = orders['Category'].str.lower()


# 4. Date-Time Conversion
orders['Order_Date'] = pd.to_datetime(orders['Order_Date'])

# 5. Merging Datasets
df = pd.merge(customers, orders, on='Customer_ID')

# 6. Feature Engineering
df['Month'] = df['Order_Date'].dt.month
# Create Total_Value flag
df['High_Value'] = df['Amount'] > 50000

# 7. GroupBy Analysis
region_sales = df.groupby('Region')['Amount'].sum()

# Total amount by Category
category_sales = df.groupby('Category')['Amount'].sum()

# 8. Pivot Reporting
pivot_table = pd.pivot_table(
    df,
    values='Amount',
    index='Region',
    columns='Category',
    aggfunc='sum'
)

print(df)
print(region_sales)
print(category_sales)
print(pivot_table)
```
