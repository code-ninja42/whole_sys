```python
import pandas as pd 
```

```python
students = pd.DataFrame({
    'Student_ID':[101,102,103,104],
    'Name':['Alice','Bob','Charlie','David'],
    'Age':[20,21,22,20]
})
```

```python
marks = pd.DataFrame({
    'Student_ID': [101, 102, 103, 105],
    'Math': [85, 90, 78, 95],
    'Science': [88, 91, 80, 96]
})

print(marks)
```

```python
city = pd.DataFrame({
    'Student_ID': [101, 102, 103, 104,105],
    'City': ['Delhi', 'Mumbai', 'Pune', 'Jaipur','Goa']
})

print(city)
```

```python
pd.concat([students,marks],axis=0)
```

```python
pd.concat([students,marks],axis=1)
```

Merge and Join

merge combines dataframes using a column's values to identify common entries 

join combines dataframes using the index to identify common entries 


```python
print(students)
```

```python
print(city)
```

```python
r1 = pd.merge(students,city,on="Student_ID",how='left')
print(r1)
```

```python
r2 = pd.merge(students,city,on='Student_ID',how='right')
print(r2)
```

```python
r3 = pd.merge(students,city,on='Student_ID',how='outer')
print(r3)
```

```python
r4 = pd.merge(students,city,on='Student_ID',how='inner')
print(r4) 
# default join
```

join() joins DataFrame using the index by default

join() works with index so first we need to make Student_ID as index

```python
students = students.set_index('Student_ID')
marks = marks.set_index('Student_ID')
r5 = students.join(marks)
print(r5)
```
