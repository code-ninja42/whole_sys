Set Operation

```python
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}
#Union(|)
s1 | s2 
#two set all element one tim print into single set 
#Intersection(&)
s1 & s2 
#common value of two sets
#Difference(-)
s1-s2 
#s1 items which are not present in s2
s2-s1
#s2 items which are not present in s1
#Symmetric Difference(^)
s1^s2
#print element of s1 and s2 and not common ele
#membership Test in not in
1 in s1
#iteration
for i in s1:
    print(i)
```

Set Functions

```python
s = {3,1,4,5,2,7}
len(s)
sum(s)
min(s)
max(s)
sorted(s) #always returns list default ascending
```

```python
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}
s1.union(s1)
s1.update(s2)
print(s1)
print(s2)
s1.intersection(s2)
s1.intersection_update(s2)
print(s1)
print(s2)
```

```python
#differnce/difference_update
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}
s1.difference(s2)
s1.difference_update(s2)
print(s1)
print(s2)
```

```python
#symmetric_difference/symmetric_difference_update
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}
s1.symmetric_difference(s2)
s1.symmetric_difference_update(s2)
print(s1)
print(s2)
```

```python
#isdisjoint/issubset/issuperset
s1 = {1,2,3,4}
s2 = {7,8,5,6}
s1.isdisjoint(s2)
#disjoint - nothing common ele in two sets

```

```python
s1 = {1,2,3,4,5}
s2 = {3,4,5}
s2.issubset(s1)
s1.issubset(s2)
s1.issuperset(s2)
s2.issuperset(s1)
#superset contains all ele of another set
```

```python
#copy
s1 = {1,2,3}
s2 = s1.copy()
#shallow
print(s1)
print(s2)
print(id(s1))
print(id(s2))
s1 = {i for i in range(1,11)}
s1
```
