```python
'''Guess game 0 to 100 jackpot number
user guess jackpot 50 upper lower'''
import random
jackpot = random.randint(1,100)
num= int(input('enter number to guess'))
counter = 1
while num != jackpot:
    if num < jackpot:
        print('guess higher')
    else:
        print('guess lowe')
    num = int(input('enter number to guess'))
    count += 1
else:
    print('correct guess')
    print('attemps:',counter)
```

```python
#write the code to count number words in a string without split()
s = input('enter the string')
l1 = []
temp = ''
for i in s:
    if i != ' ':
        temp = temp + i 
    else:
        l1.append(temp)
        temp = ''
print(l1)
```

```python
#write a python program to convert a string to title case without using the title()
s = 'Hi How are you'
L1 = []
for i in s.split():
    L1.append(i[0].upper() + i[1:].lower())
L1
print(" ".join(L1))
```
