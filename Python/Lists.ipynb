```python
#print all number divisible by 5 in the range of 1 to 50
l1 = []
for i in range(1,51):
    if i %5==0:
        l1.append(i)
print(l1)
l1 = [i for i in range(1,51) if i%5==0]
l1
```

```python
#find the lan which starts with letter p 
lan = ['java','python','php','c','javascript']
l1 = []
for i in lan:
    if i[0] == 'p':
       l1.append(i)
l1
l1 = [i for i in lan if i[0] == 'p']
l1
l1 = [i for i in lan if i.startswith('p')]
l1
```

```python
#Nested if with list Comprehension
basket = ['apple','guava','cherry','banana']
my_fruits = ['apple','kiwi','grapes','banana']
#add new list from my_fruits and items if the fruit exists in basket and also starts with 'a' 
l1 = []
for i in my_fruits:
    if i in basket and i.startswith('a'):
        l1.append(i)
print(l1)
l1 = [i for i in my_fruits if i in basket and i.startswith('a')]
l1
```

```python
#print a (3,3) matrix using list comprehension ->nested list comprehension
a = [[i*j for i in range(1,4)] for j in range(1,4)]
```

```python
#cartesian products -> list Comprehension on 2 lists together
L1 = [1,2,3,4]
L2 = [5,6,7,8]
[i*j for i in L1 for j in L2 ]
```

2 ways to traverse a list 
itemwise 
indexwise 

```python
#itemwise 
L1 = [1,2,3,4]
for i in L1:
    print(i)
```

```python
# indexwise 
for i in range(0,len(L1)):
    print(i)
    print(L1[i]) #ele
```

```python
Zip 
returns a zip object which is an iterator of tuples where the first item in each passed iterator is paired together,
and then the second item in each passed iterator are paired together 
if the passed iterators have different lengths, the iterator with the least items decides the lenth of the new iterator 
```

```python
#write a program to add items of 2 lists indexwise
L1 = [1,2,3,4]
L2 = [-1,-2,-3,-4]
zip(L1,L2) #-> zip object <zip at 0x27032014700>
list(zip(L1,L2))
[i+j for i,j in zip(L1,L2)]
```
