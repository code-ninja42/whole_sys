```python
#pop

d = {
    'name':'jack',
    'age': 26,
    3: 3
}
#d.pop(3)
d
#pop pass key it remove key and that value 
#popitem
#even though dictionaries do not support indexing 
# it works popitem cozz they internally remember insertion order works like last in first out
d.popitem()
d.popitem() # last key-value 
print(d)
del(d) #entire dict dele
del d['name'] #remove particular key value
d.clear() #empty dict
print(d)
```

```python
# update
d1 = {1:2,3:4,4:5}
d2 = {4:7,6:8}
d1.update(d2)
print(d1)
```

```python
#dictionary Comprehension
#{key:value for i in iterable}

#print 1st 10 numbers and their square
d1 = {i: i**2 for i in range(1,11)}
d1
```

```python
# using existing dict
distance = {'delhi':1000,'mumbai':2000,'banglore':3000}
{i:j*0.52 for (i,j) in distance.items()}
```

```python
# using zip
days = ["Sunday","Monday","Tuesday","Wednesday","THrusday","Satur"]
temp_c = [30.5,32.6,31.8,33.4,29.8,30.2,29.9]
{i:j for (i,j) in zip(days,temp_c)}
```

```python
# using id
products = {'phone':10,'laptop':0,'charger':0,'tablet':30}
{i:j for (i,j) in products.items() if j > 0}
```

```python
# Nested Comprehension
# print tables of number from 2 oto 4
{i:{j:i*j for j in range(1,11)}for i in range(2,5)}
```
