1. Is Python synchronous or asynchronous?

Python code is synchronous by default: it normally runs one instruction at a time, in order. Python also supports asynchronous programming using async and await, which is useful when a program spends time waiting for things like network or file operations.

2. What is a thread?

A thread is a path of execution within a program. A program can use multiple threads to handle more than one task, such as downloading a file while keeping a user interface responsive.

Single-threading: one thread executes tasks.

Multithreading: multiple threads are used within one process to make progress on tasks.

In standard CPython, the Global Interpreter Lock (GIL) means threads generally don't execute Python bytecode in parallel for CPU-heavy work. Threads can still be useful for tasks that spend time waiting, such as network requests.

3. Five common thread states

State

	

Simple meaning




New

	

Thread is created, but has not started yet.




Runnable (Ready)

	

Thread is ready to run and waiting for CPU time.




Running

	

Thread is currently executing.




Blocked / Waiting

	

Thread is waiting for something, such as input or a resource.




Terminated

	

Thread has finished its work.

Simple flow:

New → Runnable → Running → Terminated
                  ↕
           Blocked / Waiting
                  ↓
              Runnable

A running thread may become blocked while waiting. Once the event or resource it needs is available, it can return to the runnable state. These are general operating-system thread states; exact names and transitions can vary.
