An iterator in Python is an object used to traverse through all the elements of a collection -> like list,tuple, dict one element at a time

it has two key methods:
* __iter__() : Return the iterator object itself 
* __next__() : Returns the next value from the sequence

Need For Iterators:
* it Processes items only when needed it do not store entire item at memory like sequential data types saving memory


Iterable 
* An iterable is any object that can return an iterator (like list tuple)
