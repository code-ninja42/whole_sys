## Polymorphism

class Dog(Animal):
    def sound(self):

## Abstraction 
Abstraction means hiding implementation details and showing only essential features


from abc import ABC, abstractmethod

class Animal(ABC):

    @abstractmethod
    def sound(self):
        pass


class Dog(Animal):

    def sound(self):
        print("Bark")


d = Dog()
d.sound()


Abstraction hides implementation details and exposes only the required interface. In Python, we can use ABC and @abstractmethod to create abstract classes and abstract methods. A concrete child class must implement the abstract methods before its object can be created.

Exception Handling

An exception is an abnormal condition that occurs during program execution and can be handled by the program.

try:
    x = 10 / 0
except ZeroDivisionError:
    print("Cannot divide by zero")
finally:
    print("Done")


try

Contains code that might raise an exception.

except

Handles the exception.

finally

Runs whether an exception occurs or not.
