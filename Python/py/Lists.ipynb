## List 
A list is an ordered, mutable collection of elements in Python. it stores elements of different data types


* Ordered: Elements maintain their position and insertion order
* Mutable: Elements can be added,removed or modified after creation
* Allows duplicates: The same value can appear multiple times
* Heterogeneous: A list can contain different data types
