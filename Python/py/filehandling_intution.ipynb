File Handling

file handling in Python is the process of creating,reading,writing,updating and deleting files using built-in functions or methods ,file handling allows us to store data permanently, even after the program has stopped running

Mode

	

Meaning

"r"

	

Read a file; error if it does not exist

"w"

	

Write to a file; creates it or overwrites existing content

"a"

	

Append data to the end; creates file if missing

"x"

	

Create a new file; error if it already exists

"r+"

	

Read and write; file must exist

"w+"

	

Read and write; overwrites existing content or creates file

"a+"

	

Read and append; creates file if missing

"rb"

	

Read in binary mode

"wb"

	

Write in binary mode
