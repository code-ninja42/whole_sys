Generator - Function

Python generators are simple way of creating iterators

```python
def gen_demo():
    yield "first statement"
    yield "second statement"
    yield "third statement"    
```

```python
gen = gen_demo()
print(gen) # return generator object
print(next(gen))
print(next(gen))
```

yield and return 

yield will execute when generator object create and called unlike return it execute when fucntion called

generator will keep memory 

normal function won't

return : Returns a value and ends function 

yield : Produces a value and pause execution
