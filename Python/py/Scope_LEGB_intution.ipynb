## Function Scope

Scope refers to the region of a program where a variable can be accessed 

1.Local

2.Enclosing

3.Global

4.Built-in

A.Local Scope

A variable created inside a function is local and can normally be accessed only within that function

def greet():
    message = "Hello"
    print(message)

greet() # like this access it
print(message) # 3rror

B.Global

created outside function.it can be accesse can be accessed anywhere

if u want to assign inside function, declare it with global.

C.Enclosing Scope

A variable defined in an outer function can be accessed by an inner function

def greet():
    message = "Hello"
    print(message)

greet()
print(message)

D.nonlocal

allows an inner function to reassign a variable in its nearest enclosing function scope

def outer():
    x = 20

    def inner():
        nonlocal x
        x = 50

    inner()
    print(x)

outer()

Output:
 50
