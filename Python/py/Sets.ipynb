Set Operations

1.Union 

Union returns all unique elements that are present in either set or both


2.Intersection
returns only the elements that are common to both sets


3.Difference
returns the elements present in first set but not in second set


4.Symmetric Difference

Symmetric difference returns elements that are in either set but not in both
