Shallow Copy : 

* creates a new outer object but shares the nested objects with the origiinal 
so if changes in original then change in copy to 
copy.copy() for shallow copy 

Deep copy 
* creates a completely independent copy, including all nested objects

so means changing original do not change copy one 

copy.deepcopy()

Shallow copy creates a new outer object, but nested objects are shared between the original and copy. Deep copy creates a completely independent copy, including nested objects. In Python, we can use copy.copy() for shallow copy and copy.deepcopy() for deep copy.

```python
# Shallow copy
import copy

a = [[1, 2], [3, 4]]
b = copy.copy(a)

b[0][0] = 100

print(a)
print(b)

[[100, 2], [3, 4]]
[[100, 2], [3, 4]]
```

```python
# Deep copy
import copy



a = [[1, 2], [3, 4]]

b = copy.deepcopy(a)



b[0][0] = 100



print(a)

print(b)[[1, 2], [3, 4]]

[[100, 2], [3, 4]]
```
