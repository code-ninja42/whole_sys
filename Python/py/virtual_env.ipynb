## PVM
Python virtual machines.it is the runtime component of python that execute python bytecode

When we write python program computer cannot directly execute the python source code so python first compiles the code into bytecode and then the PVM executes that bytecode

## Virtual Environment
it is isolated Python environment that allows us to install and manage package separately for each project

for example 
* project A requires jango 4
* project b requires jango 5

if the both projects use the same Python environment, installing or upgrading jango for one project could affect the other. Virtual environments help prevent this by keeping their dependencies separate
