Python manages memory automatically. You usually do not need to allocate or free memory by yourself

1.memory allocation:- when you create object python gets memory for it 

Garbage collection: when an object is no longer reachable or needed python can reclaim its memory .python uses reference counting or garbage collector to help clean up objects

a = 10
b = 10

Python may let both variables refer to the same integer object 10 in memory, rather than creating two separate objects.

a and b are variable names (references).

10 is the integer object.

When you use a or b, Python accesses that object.

“Python handles memory management automatically using private heap, reference counting, and garbage collection. When we create an object, Python allocates memory for it. Python keeps track of the object's reference count. When the reference count becomes zero, Python can deallocate that memory. For example, if a = [1,2,3] and b = a, the reference count is 2. When we del a, the reference count becomes 1. When we del b, the reference count becomes 0, so Python can deallocate that object. Python's garbage collector is also used to handle cyclic references.”
