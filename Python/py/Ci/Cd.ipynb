### What is ci/cd


CI = continous Integration

Whenever developers puch code the system automatically

code -> Build -> Test -> Report 

so the goal here is to catch problems before merging/deploying


CD = Continous Delivery/ Deployment

After the code passes CI checks, it can be prepared for deployment environment



Developer Pushes code

    |
    
    Install dependencies

    |

    static checks

    |

    Unit / integration /end-to-end testing

    |

    Build

    Deploy

smoke = everything working properly

Is the application basically working?

santiy = after updation everything working properly

Does this particular change work?
