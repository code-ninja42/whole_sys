```python
import mysql.connector
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.messages import HumanMessage, AIMessage
from langchain.agents import create_agent
```

```python
load_dotenv()
```

```python
db = mysql.connector.connect(
    host="localhost",
    user="root",
    password="",
    database=" data_science",
    port=3306
)
```

```python
llm = ChatGoogleGenerativeAI(
    model="gemini-3.6-flash",
    temperature=0
)
```

```python
from langchain_core.tools import tool

@tool
def get_student_data():
    """Get all student data from MySQL database."""
    
    cursor = db.cursor()
    cursor.execute("SELECT * FROM student_info_ai")
    
    data = cursor.fetchall()
    cursor.close()
    
    return data
```

```python
agent = create_agent(
    model=llm,
    tools=[get_student_data],
    system_prompt="""
    You are a helpful student assistant.

    Use the student database tool whenever
    the user asks about students.

    Do not make up student information.
    Use the database data to answer.
    """
)
```

```python
question = input("Ask your question: ")
```

```python
result = agent.invoke({
    "messages":[
        ("user",question)
    ]
})
print(result)
```
