```python
from dotenv import load_dotenv
import os

load_dotenv()

api_key = os.getenv("GOOGLE_API_KEY")
print(api_key is not None)
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.messages import HumanMessage
from langchain_core.prompts import PromptTemplate
```

```python
load_dotenv()
```

```python
llm = ChatGoogleGenerativeAI(
    model="gemini-3.6-flash",
    google_api_key=api_key
)
```

```python
tech = PromptTemplate(
    input_variables=["tech"],
    template="""
You are an experienced technical interviewer.

Generate 5 interview questions for {tech}.

Requirements:
1. Generate 3 basic questions.
2. Generate 2 advanced questions.
3. Keep the questions clear and practical.
"""
)
```

```python
technology = input("Enter your technology: ")
```

```python
prompt = tech.format(tech=technology)
```

```python
response = llm.invoke(prompt)
```

```python
print(response.content)
```
