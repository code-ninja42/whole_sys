```python
import mysql.connector
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.messages import HumanMessage,AIMessage
from langchain.agents import create_agent
```

```python
load_dotenv()
```

```python
def get_db_connection():
    return mysql.connector.connect(
        host="localhost",
        user="root",
        password="",
        database="data_science",
        port=3306
    )
```

```python
db_test = None

try:
    db_test = get_db_connection()

    if db_test.is_connected():
        print("MySQL connection successful!")

except mysql.connector.Error as e:
    print("MySQL connection failed:")
    print(e)

finally:
    if db_test is not None and db_test.is_connected():
        db_test.close()
        print("Connection closed.")
```

```python
@tool
def get_student_data():
    """
    Get all student records from the student_info_ai table.
    """

    db = None
    cursor = None

    try:
        db = get_db_connection()
        cursor = db.cursor()

        cursor.execute("SELECT * FROM student_info_ai")

        data = cursor.fetchall()

        return data

    except mysql.connector.Error as e:
        return f"MySQL error while getting student data: {e}"

    finally:
        if cursor is not None:
            cursor.close()

        if db is not None and db.is_connected():
            db.close()
```

```python
llm = ChatGoogleGenerativeAI(
    model="gemini-3.6-flash",
    temperature=0
)
```

```python
tools = [
    get_student_data,
    get_employee_data,
    get_department_data,
    get_product_data,
    get_order_data
]
```

```python
agent = create_agent(
    model=llm,
    tools=tools,
    system_prompt="""
You are a database assistant.

You answer questions using data retrieved from the MySQL database.

You have access to these tools:

1. get_student_data
   Use this for questions about students, marks, courses, semesters, etc.

2. get_employee_data
   Use this for questions about employees.

3. get_department_data
   Use this for questions about departments.

4. get_product_data
   Use this for questions about products.

5. get_order_data
   Use this for questions about orders.

Important rules:

- Always use the appropriate database tool when the user asks about database information.
- Do not invent database information.
- Do not assume information that is not present in the tool result.
- Analyze the returned database records and answer the user's question clearly.
- If the database does not contain the requested information, say so.
"""
)

print("Agent created successfully.")
```

```python
question = input("Ask your question: ")

result = agent.invoke({
    "messages": [
        ("user", question)
    ]
})

print(result["messages"][-1].content)
```
