## RAG
Retrival-Augmented Generation

it is a technique that combines informatiom retrieval with a LLM. RAg retrieves relevant information from an external knowledge source and provides that information to the LLM as context to generate a more accurate and grounded answer

RAG works
1. Collect documents


2.Document loading 
document loader reads the document and extracts its text

3.Text chunking

Large documents are divided into smaller pieces called chunks

- Chunking Size:-

is the maximum amount of text that we divide a document it determines how much information each individual chunk contains

- Chunking overlap:-

chunk overlap is the amount of text that is repeated between two consecutive chunks . it helps preserve context when an important sentence or idea is located near a chunk boundary 
like

Chunk 1 has:
"Employees can apply for leave after completing..."

chunk 2:
"...six months of service..."

so the second chunk retains some context from the first

without overlap the meaning can be separated
4.Create embeddings

Each chunk is converted into a numerical representation called an embedding

5.store embedding in a vector db like chroma Faiss


now  when user anl que The same embedding model converts the user's question into a vector 

The query vector is compared with vectors that are stored in vector db

The system looks for chunks whose meaning is most similar to the question and This is commonly done using measures such as cosine similarity or other vector - distance methods

The most relevant chunks are returns 

Now the LLM receives both the user's question and the retrieved information

and now LLM uses the retrieved context to generate the response
