## Langchain
Langchain is an open-source framework used to build applications powered by LLMs. It helps us to connect an LLm with prompts,external tools,APIs,db and memory to develop AI agents and RAG systems

Components Of Langchain:-

1.LLM

The model understands the user's query and generates a response or determines what action is needed
ex.openAI models,ChatGroq,Gemini

2.Prompt Templates

they structure instructions and user input before sending them to the model

3.Tools

tools allows an AI agents to perform actions using functions,APIs,db or tother services.

4.Chains

chains connect multiple processing step each step output is input for an9ther step

5.Agents

Agents use the model to decide which available tool or action is appropriate for a task. They can perform multiple steps when necessary

6.Memory 

conversation context 


### Simple Langchain + Tool example

from langchain_groq import ChatGroq
from langchain.tools import tool
from langchain.agents import create_agent


# 1. Create a tool
@tool
def get_product_price(product: str) -> str:
    """Get the price of a product."""
    
    products = {
        "laptop": 50000,
        "phone": 20000,
        "headphone": 3000
    }

    return str(products.get(product.lower(), "Product not found"))


# 2. Create LLM
llm = ChatGroq(
    model="llama-3.3-70b-versatile",
    temperature=0
)


# 3. Create Agent
agent = create_agent(
    model=llm,
    tools=[get_product_price]
)


# 4. Ask the agent a question
response = agent.invoke({
    "messages": [
        {
            "role": "user",
            "content": "What is the price of the laptop?"
        }
    ]
})


# 5. Print final response
print(response["messages"][-1].content)

Gemini model

from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate

# Gemini model
llm = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash",
    temperature=0
)

# Prompt
prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful AI assistant."),
    ("human", "{question}")
])

# Create chain
chain = prompt | llm

# User input
question = input("You: ")

# Get response
response = chain.invoke({
    "question": question
})

print("AI:", response.content)
