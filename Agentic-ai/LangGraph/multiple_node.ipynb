```python
from typing import TypedDict
from langgraph.graph import StateGraph, START, END
from langchain_google_genai import ChatGoogleGenerativeAI
```

```python
class State(TypedDict):
    question:str
    answer:str
```

```python
llm = ChatGoogleGenerativeAI(
    model = "gemini-3.5-flash",
    temperature=0
)
```

```python
# Question Node
def question_node(state: State):
    print("Question Node Running...")
    
    return {
        "question": state["question"]
    }
```

```python
# LLM Node
def llm_node(state: State):
    print("LLM Node Running...")
    
    response = llm.invoke(state["question"])
    
    return {
        "answer": response.content
    }
```

```python
# Response Node
def response_node(state: State):
    print("Response Node Running...")
    print("Question:", state["question"])
    print("Answer:", state["answer"])
    
    return {
        "question": state["question"],
        "answer": state["answer"]
    }
```

```python
graph = StateGraph(State)
```

```python
graph.add_node("question",question_node)
graph.add_node("llm",llm_node)
graph.add_node("response",response_node)
```

```python
graph.add_edge(START,"question")
graph.add_edge("question","llm")
graph.add_edge("llm","response")
graph.add_edge("response",END)
```

```python
app = graph.compile()
```

```python
result = app.invoke({
    "question": "What is Programming?",
    "answer": ""
})


print("\nFinal Answer:")
print(result["answer"])
```
