## CrewAI 
crewAI is a python framework for building multi-agent it allows us to create multiple AI agents where each agent has specific role goal task assign go them and coordinate their work to complete a large objective

### Main components
1.Agent
An agent is an AI worker with a specific role , goal and capabilities

An agent can also be given tools to interect with external systems

2.Task

A task defines the work that an agent needs to perform

3.Crew 
A crew is the team that contains agents and tasks

4.Process
defines how tasks are executed

Sequential:-
Hierarchical:-
A manager coordinates the work of other agents
front agent

5.Tools
allows agents connect with external sources



Backstory:-
it is the background or context given to an AI agent that describes its experience expertise and personality 

Role = Who you are
Goal = What you need to achieve
Backstory = What experience/context you have

Kickoff() is the execution method of CrewAi. it starts the crew's workflow and returns the result after the assigned tasks are completed

import os
from dotenv import load_dotenv

from crewai import Agent, Task, Crew, Process, LLM

load_dotenv()

# Gemini LLM
llm = LLM(
    model="gemini/gemini-2.5-flash",
    api_key=os.getenv("GOOGLE_API_KEY")
)

# 1. Research Agent
researcher = Agent(
    role="Researcher",
    goal="Collect accurate information about the given topic",
    backstory="You are an experienced researcher who finds and organizes useful information.",
    llm=llm
)

# 2. Analyst Agent
analyst = Agent(
    role="Data Analyst",
    goal="Analyze the research information and identify important insights",
    backstory="You are an experienced analyst who converts raw information into useful insights.",
    llm=llm
)

# 3. Writer Agent
writer = Agent(
    role="Technical Writer",
    goal="Create a clear final report from the research and analysis",
    backstory="You are an experienced technical writer who explains complex topics clearly.",
    llm=llm
)

# Tasks
research_task = Task(
    description="Research the latest developments in Generative AI.",
    expected_output="A structured summary of important Generative AI developments.",
    agent=researcher
)

analysis_task = Task(
    description="Analyze the research and identify the most important points.",
    expected_output="A clear analysis with important insights.",
    agent=analyst
)

writing_task = Task(
    description="Create a final report using the research and analysis.",
    expected_output="A well-structured final report.",
    agent=writer
)

# Main/Manager Agent
manager = Agent(
    role="Project Manager",
    goal="Coordinate the agents and ensure the final report is accurate and complete.",
    backstory="You are an experienced project manager who coordinates AI agents and reviews their work.",
    llm=llm
)

# Crew
crew = Crew(
    agents=[researcher, analyst, writer],
    tasks=[research_task, analysis_task, writing_task],
    manager_agent=manager,
    process=Process.hierarchical,
    verbose=True
)

# Start workflow
result = crew.kickoff()

print("\nFinal Result:")
print(result)
