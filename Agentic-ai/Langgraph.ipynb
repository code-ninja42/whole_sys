### LangGraph
it is a framework that used to build stateful,multi-step and controllable AI agents workflows. it represent AI workflow as a graph where different steps are connected as nodes and edges 

where node perfrom task and edges connect this nodes

## Main Components

1.State

state stores the information that moves through the worklflow


2.Nodes

Nodes are individual step or function that perform specific tasks


3.Edges 

Edges define the connection and flow between nodes

4. start and end

5.Checkpoint/ Persistence

LangGraph can save graph state through checkpointers, allowing workflows to maintain state across steps and depending on the setup across interaction 

This is useful for:-
 * Human approval steps or 
 * conversation steps

import os
from dotenv import load_dotenv

from typing import TypedDict
from langchain_google_genai import ChatGoogleGenerativeAI
from langgraph.graph import StateGraph, START, END


# Load .env
load_dotenv()

# Gemini model
llm = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash",
    google_api_key=os.getenv("GOOGLE_API_KEY"),
    temperature=0
)


# 1. Define State
class State(TypedDict):
    question: str
    answer: str


# 2. Define Node
def ask_gemini(state: State):
    response = llm.invoke(state["question"])

    return {
        "answer": response.content
    }


# 3. Create Graph
graph = StateGraph(State)


# 4. Add Node
graph.add_node("ask_gemini", ask_gemini)


# 5. Add Edges
graph.add_edge(START, "ask_gemini")
graph.add_edge("ask_gemini", END)


# 6. Compile Graph
app = graph.compile()


# 7. User Input
question = input("You: ")


# 8. Run Graph
result = app.invoke({
    "question": question,
    "answer": ""
})


print("AI:", result["answer"])

Compile converts the graph definition into an executable application. Before compilation, we define the nodes, edges, and workflow. After compilation, we can run the graph using invoke()

#### When to use langchain:-

Use a chain when the workflow is linear and follow the same sequence every time

- Framework for building LLM applications

- Useful for chains,prompts,tools,retrieval
- task
#### When to use graph :- 

Use Langgraph when the workflow has multiple steps, decisions , loops and states

- Framework for building stateful agent workflows

- useful for complex multi-step workflows

- nodes
