LSTM - Long Short Term Memory


LSTM is a type of RNN designed to handle long-term dependencies
and reduce the vanishing-gradient problem. it has two types of states Cell State and Hidden State and three main gates Forgot Gate,Input Gate and Output Gate

### Forgot Gate 
The Forgot Gate decides which information from the previous Cell State should be kept and which should be forgotten.

it takes
xt + ht-1

and produces values between 0 to 1 using sigmoid.

### Input Gate
input gate decides how much new information should be added to the memory

### Candidate Cell State - Along with input Gate
The Candidate cell state creates the actual new information that could be stored

it uses tanh, giving values from -1 to 1


### Cell State - Memory Is Updated 

now lstm have

Useful old memory
     + 
Useful new information
        +
New Cell State C(t)


Cell State = long-term memory

it carries important information forward to future time steps

### Output Gate 
Now the LSTM has updated its memory

After updating the cell state, the output gate decides which part of the current memory should be exposed as the output

### Hidden State - LAST
Hidden state = current output/short-term information

"What is the difference between cell state and hidden state?"

Say:

"The cell state represents long-term memory and carries important information across time steps, while the hidden state represents the current output or short-term information and is passed to the next time step."

in short 
Input Gate -> How much new information should I add?
Candidate -> What new information can I add
