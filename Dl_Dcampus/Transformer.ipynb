# Transformer
neural network architecture designed mainly for sequence data, especially NLP. rnn and lstm process sequences sequentially which takes long training time

A Transformer is a neural network architecture that uses self-attention to understand relationships between words and process sequences efficiently.

Self-attention allows each word in a sentence to look at other words in the same sentence and determine how important they are to understanding its meaning.

Attention allows a model to focus more on the important words when processing a sequence.

Main ides of Transformer 

The Transformer says:

Instead of processing words one by one , look at the all words together and determine which words are important to each other using attention
