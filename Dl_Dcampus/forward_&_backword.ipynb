**Forward Propagation:**
Forward propagation is the process of passing input data through the neural network from the input layer to the output layer to generate a prediction.

At each neuron, we calculate:

$$
z = wx+b
$$

Then we apply an activation function:

$$
a=f(z)
$$

This process continues through the layers until we get the final prediction. Then we calculate the loss by comparing the prediction with the actual value.

**Backpropagation:**
After calculating the loss, backpropagation works backward through the network. It calculates how much each weight and bias contributed to the loss by calculating gradients.

These gradients are calculated using the **chain rule of differentiation**.

Finally, the optimizer uses these gradients and the learning rate to update the weights:

$$
w_{new}=w-\eta\frac{dL}{dw}
$$

The goal is to reduce the loss and improve the model's predictions.

**In simple terms:**
Forward propagation → **make prediction**
Loss function → **measure error**
Backpropagation → **find how weights caused the error**
Optimizer → **update the weights**



What is Chain Rule?

The chain rule is a mathematical method used to calculate how a change in one variable affects another variable when there are multiple steps between them.

In a neural network, a weight affects the neuron output, the neuron output affects the prediction, and the prediction affects the loss.

There are three common approaches based on how much training data is used for each update. Batch Gradient Descent uses the entire dataset for one update. Stochastic Gradient Descent uses one training sample for each update. Mini-Batch Gradient Descent uses a small group of samples, such as 32 or 64, for each update. Mini-batch is commonly used in deep learning because it provides a balance between computational efficiency and stable gradient updates.
