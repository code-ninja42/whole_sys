### GRU = Gated Recurrent Unit 

GRU was introduced as a simpler alternative to LSTM

LSTM works well for long-term dependencies, but its architecture is realtively complex:
3 gate 
cell state 
hidden state 
Candidate cell state 
More computations
More parameters

So GRU was designed to provide similar benefits with a simpler architecture and fewer parameters

The main difference is complexity

GRU Architecture has 

2 gates

1.Update Gate

2.Reset Gate

GRU uses only:

Hidden State

#### Reset Gate
How much of the previous hidden state should be used when creating new information?

#### Update Gate
How much old information should i keep and how much new information should i use

#### Candidate Hidden State
it uses: 
    Current input + Reset-filtered previous hidden state

##### Hidden State
GRU combines the old hidden state and candidate hidden state

#### Easy Example
