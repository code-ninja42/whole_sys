## VAE - Variational Autoencoder 

it is type of generative deep learning model that learns the underlying distribution of data and can generate new data similar to the training data


#### 1. Encoder
The encoder takes the input and compresses it into a latent representation 
unlike a normal autoencoder vae does not directly produce one fixed latent vector 

It produces:
mean 
std
These describe a probability distribution

latent vector stored at latent space 

### 2.Decoder 
takes the sampled latent vector z and tries to reconstruct the original output


#### Latent Vector
A latent vector is the compressed numerical representation of one particular input 

and
#### Latent Space 
Latent space is the entire space in which these latent vectors are stored
