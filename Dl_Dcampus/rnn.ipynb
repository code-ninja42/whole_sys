### RNN 
recurrent Neural Network it is a type of neural network designed for sequential or time-depedent data. where the order of the data matters ex.:- text,speech,stock price , time-series analysis 

RNN has hidden state which acts like memory at each time step the RNN takes the current input along with information from the previous time step and produces an output and an updated hidden state

hidden state is memory of RNN


ex :- I -> do -> machine -> learning

step 1:- input - I

There is no previous information initially so 
previous hidden state = h0 

New hidden state = h1

"I" -> RNN -> h1

step 2 :- input - do
h1 has I 
so now h2 contain some useful information from :-
I + do 
so on 

So the hidden state is continously updated as the sequence is processed

RNN Formula 

    ht = f(wx.Xt + Wh.ht-1 + b)

xt = input 
ht-1 = previous hidden state/memory
wx = weights for current input 
wh = weights for previous memory
ht = new hidden state


BPTT - BackPropagation Through Time
need to update weights so now Loss -> h4 ->h3 -> h2 ->h1 this is called BPTT

### Vanishing Gradient Problem
During BPTT, gradient are passed backward through many time steps 
and these value keeps becoming smaller

as the effect 
 RNN struggles to learn long-term dependencies

### Exploding Gradient Problem

During BPTT, gradient becomes extremely large
as the effect
make training unstable
