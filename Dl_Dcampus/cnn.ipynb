Cnn - convolutional Neural Network,is a deep learning architecture mainly used for image and computer vision tasks. The main idea of CNN is that instead of manually extracting features from an image, it automatically learns important features such as edges,textures,shapes and eventually complex objects 

### Cnn Architecture

first the input layer takes image in the form of pixel. if 224x224 when color image can have dimension like 224x224x3, where 3 represents the RGB channels

Now at convolution layer. Here we use small filter/kernerls  

filter or kernel :- is a small matrix of learnable weights that used to detect a particular feature in an image, such as an edge corner or texture 
now it start at top-left corner and take small 3x3 matrix and multiply then Those filter slides across the image from one position to another and cover whole image
convolution and filter or kernerls multiply and return feature map

After Convolution layer we generally apply an activation function such as ReLu(why ReLU here if negative value so it do 0 that's why if pos then return that value).activation introduce non-linearity and helps the network learn complex patterns

Then we can use Pooling layer, 2 type max and avg 

Pooling Layer :- reduce the spatial dimensions of the feature maps.reduces computaion and helps the model focus on important features 

feature map = [ 2 8 
                7 6    ]   apply max pooling = 7 

These convolution,activation, and pooling operations can be repeated multiple times to cover all features of image

at flatten layer this feauture map conver into one -dimension vector. Then This vector passed to fully connected layer

at Fully connected layer  Dense  are connected through layer , input layer multiple hidden layer and one output layer , the output layer depends on the problem. For binary classification we generally use sigmoid for multi-class classification we generally use multiple output neurons with softmax activation
