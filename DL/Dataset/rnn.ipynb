```python
import numpy as np 
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding,SimpleRNN,Dense
```

```python
X = np.array([
    [1,2],
    [3,4],
    [5,6],
    [7,8]
])
y = np.array([
    9,
    9,
    10,
    11
])
```

```python
words = {
    1:"I",
    2:"love",
    3:"Virat",
    4:"plays",
    5:"MS",
    6:"Dhoni",
    7:"We",
    8:"watch",
    9:"cricket",
    10:"captain",
    11:"football"
}
```

```python
model = Sequential()
model.add(Embedding(input_dim=12,output_dim=8))
model.add(SimpleRNN(16)) #hidden layer
model.add(Dense(12,activation="softmax"))
#12 input dims 
```

```python
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)
```

```python
model.fit(X,y,epochs=300)
print("Training Completed")
```

```python
test = np.array([[1,2]])
prediction = model.predict(test)
index = np.argmax(prediction)
print(index)
```

```python
print("Next word:",words[index])
```
