## Logistic Regression

first multilinear regression line
z = w0 + w1.x1 + w2.x2

where 
w0 = bias 
w1 , w2 = weight 
x1 x2 = input feature

intially the algorithm starts with small random values or zeros

during the training, the model makes predictions calculate the error using the loss function 

and then updates both the weights and the bias using gradient descent 

This process repeats until the loss is minimized

How are w0 and w1 determined in logistic regression ?

w1 is the weight of feature x1 . it tells us how much the linear score z changes when x1 increases by one unit, while keeping all other features constant

w0 is the  bias or intercept , it is the baseline value of z before considering any input features. when all feature values are zero 
then the value of z 

e.g. Imagine 

base salary = 30k
overtime = 500 
bonus = 1000

even if you do 0 hours of overtime and 0 projects you still earn 30k that's exactly what the bias does

The bias(w0) is the intercept or baseline value of the model.it determines the initial value of z when all features are zero,while the weights (w1,w2,...)determine how much z changes when their corresponding features increase by one unit. Both the bias and the weights are learned automatically during training by minimizing the loss function

## Why sigmoid 

we don't use only the value of z because z can be any real number,(-15,30,1000) which is difficult to interpret as probability.The sigmoid function convert z into a value between 0 and 1 , allowing us to interpret it as the probability 

for  e.g z = 0.80 it means: 
* 80% probability of being selected (Class =1)
* 20% probability of not being selected(Class =2)

Binary - Cross -Entropy
Binary cross-entropy (also called log loss) measures how far predicted probabilities are from the true labels. Low loss means confident and correct. high loss penalizes confident but wrong predictions a lot. We minimize it during training, often with gradient descent, to update the weights step by step. So, when loss is high, gradients push updates to reduce that loss. That's how the model gradually learns better probability estimates.
