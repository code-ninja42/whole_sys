PCA - Principal Component Analysis 
it is dimensionality reduction technique that helps to reduce the number of features in a dataset while keeping the most important information

it uses linear algebra to transform data into new features called principal componenets.
it find these by calculating 

eigenvectors -> directions 
and eigenvalues -> importance 
from the covariance matrix


Step 1:- scaling 
different features may have different units and scales e.g salary vs age 

PCA first standardizes the data by making each feature have :
* mean 0 
* a std 1 

Step 2 :- Calculate Covariance Matrix 
next PCA calculates the covariance matrix to see how features relate to each other whether they incr or de together 

Step 3 :- Find the Principal Components 
* 1st Principal Component (PC1): The direction of maximum variance (most spread).

* 2nd Principal Component (PC2): The next best direction, perpendicular to PC1 and so on.
