What is Random Forest?

Random Forest is an ensemble learning algorithm that builds multiple decision trees and combines their predictions to produce a more accurate and stable result.

For classification → majority voting
For regression → average of predictions

2. What is Bootstrap?

Suppose your training dataset has 1,000 rows.

Random Forest creates a different dataset for each tree by doing bootstrap sampling:

Randomly select 1,000 rows with replacement.
Because sampling is with replacement, some rows can appear multiple times.
Some original rows will not be selected at all.
This creates a different training sample for each tree.

Example:

Original:

A B C D E

Bootstrap sample:

A C C D E

Here, C appears twice and B is not selected.

This process is called Bootstrap Aggregating (Bagging).

3. What happens with the rows not selected?

The rows that are not included in a tree's bootstrap sample are called Out-of-Bag (OOB) samples.

Random Forest can use these OOB samples to evaluate that tree's performance without needing a separate validation set for that purpose.

4. Random feature selection

Random Forest adds another source of randomness.

When a tree needs to split a node, it doesn't necessarily consider all features. It randomly selects a subset of features and finds the best split among them.

For example:

You have 10 features:

Age, Salary, Experience, BMI, Height, Weight, ...

A particular split might consider only 3 randomly selected features.

So there are two types of randomness:

Random rows → Bootstrap sampling
Random features → Feature subset at each split

5. How are predictions combined?

Suppose Random Forest has 5 trees:

Classification:

Tree 1 → Failure
Tree 2 → Failure
Tree 3 → No Failure
Tree 4 → Failure
Tree 5 → No Failure

Majority vote → Failure

For regression:

Tree predictions = 50, 55, 52, 48, 51

Final prediction ≈ 51.2

Why does this reduce overfitting?

A single decision tree can easily become very complex and overfit the training data.

Random Forest creates many different trees using different samples and features. Their individual errors are less likely to be identical, and combining them makes the final model more stable and less sensitive to noise.

## XGBoost
XGBoost

XGBoost (Extreme Gradient Boosting) is a supervised machine learning algorithm based on Gradient Boosting. It combines multiple Decision Trees sequentially to build a strong model.

How it works
Start with a simple prediction.
Build the first Decision Tree.
Calculate the errors (residuals) of the current predictions.
Build the next tree to focus on correcting those errors.
Repeat this process for multiple trees.
Each new tree is added to improve the previous model.
The predictions from all trees are combined to get the final prediction.
