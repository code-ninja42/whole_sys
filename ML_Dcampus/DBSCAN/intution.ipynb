DBSCAN Clustering

Density - Based Spatial clustering of Application with Noise 

it groups points that are close together and have many nearby points 

it can also identify outliers/noise

DBSCAN adv:
it does not required pre define k number 
it can find any clusters shape 
it can detect outlier

Important Parameters

1.eps
means the maximum distance to consider another point as a neighbor

Think of it as a radius around a point 

small eps -> fewer neighbors -> many points may become noise 

Large eps -> more neighbors -> different clusters may merge 


2.MinPts 
minimum number of nearby points needed to consider an area dense



## Three Types of Points
DBSCAN classifies points into 3 types 

1.Core point 
A point having at least Minpts points within eps distance 

it is the main point used to form a cluster 

2.Border Point
A point that is near a core point, but doen not have enough neighbors to considered as core point itself

3.Noise point
point that is not connect to any cluster 



How DBSCAN Works

step 1. choose a point 
step 2. check its neighbors
it looks at points within the eps distance 
step 3. check MinPts 
* Enough neigbhors -> Core point 
step 4: create cluster
Starting from a core point, DBSCAN adds nearby connected points to the same cluster 
step 5. expanding the cluster through other core points
