MAE is robust to outliers because it uses the absolute value of errors instead of squaring them. As a result, each error contributes proportionally to the final metric. In contrast, MSE squares the errors, causing large errors (outliers) to have a disproportionately large impact. Therefore, MAE is less sensitive to outliers and is considered more robust.

Assumptions 

Sure. For linear regression, we usually assume a straight line relationship between the inputs and the target. We want the errors to be independent, not showing patterns over time or across observations. And we don't want the inputs to be highly correlated with each other, because that makes it hard to separate their individual effects. Also, we assume constant variance of errors. So the spread of residuals should look roughly the same for all predicted values. If that spread fans out or shrinks, that's a violation. And lastly, we often assume residuals are approximately normal, which helps with hypothesis tests and confidence intervals.


Asssumption of Linear Regression

* Linearity – There should be a linear relationship between independent and dependent variables.
* Independence of Errors – Errors should be independent of each other.
* Homoscedasticity – Error variance should remain constant.
* Normality of Residuals – Residuals should be approximately normally distributed.
* No Multicollinearity – Independent variables should not be highly correlated with each other.
* No Significant Outliers – Extreme outliers should not strongly influence the model.

Residual

Residual is the error for one data point:

Residual = Actual − Predicted

Example:

Actual = 10, Predicted = 8

Residual = 10 − 8 = 2

Loss Function

A loss function measures the overall error of the model using one or more residuals.

For Linear Regression, commonly:

MSE = Average of squared residuals

Loss = individual error
Cost = average/total loss for the dataset
