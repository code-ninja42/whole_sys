Overfitting occurs when a model learns the training data too closely, including noise and unnecessary patterns. As a result, it performs very well on training data but poorly on unseen/test data.

Ways to Reduce Overfitting

1. Feature Engineering

Remove irrelevant or noisy features.
Create meaningful features.
Reduce unnecessary complexity in the input data.

2. Feature Scaling

Important for distance-based algorithms such as KNN and K-Means.
It prevents features with larger numerical ranges from dominating the distance calculation.

3. Pruning — Tree-based Models

Remove unnecessary branches from a Decision Tree.
This makes the tree simpler and reduces overfitting.

4. Limit max_depth — Tree-based Models

Set a maximum depth for the tree.
A shallow tree is less likely to memorize the training data.

5. Regularization — L1 and L2

Adds a penalty for large model weights.
L1 (Lasso) can make some weights exactly zero, helping with feature selection.
L2 (Ridge) makes weights smaller and reduces model complexity.

6. PCA

Reduces the number of features by converting them into fewer principal components.
This can reduce noise and model complexity.

7. Dropout — Deep Learning

Randomly turns off some neurons during training.
This prevents the network from depending too much on specific neurons.

8. Early Stopping — Deep Learning

Monitor validation performance during training.
Stop training when validation performance stops improving or starts getting worse.

This prevents the model from learning the training data too much
