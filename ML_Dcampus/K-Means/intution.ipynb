## K-Means

K-means is unsupervised techniq used to divide data into k groups based on similarity 

it groups points that are close to each other 

K - represent the number of clusters we want

Centroid - mean/avg position of points in a cluster

K-means working

step 1: Choose K 
Decide how many cluster 


step 2: intialize centroids 
Randomly select K points as initial cluster centroids

step 3: assign points to nearest centroid
Assign each data point to the nearest centroid using
distance, usually Euclidean distance

Each point goes to the nearest centroid 

step 4: Update centroids 
After assiging points, calculate the avg of all points in each cluster 
That avg become the new centroid

step 5: Repeat 
this continues until
* centroid stop changing

WCSS in K-Means

WCSS (Within-Cluster Sum of Squares) measures how close the data points are to their own cluster centroid.

For each cluster:

Calculate the distance between every point and its centroid.
Square each distance.
Add all the squared distances.

Formula:

$$ WCSS = \sum_{i=1}^{K}\sum_{x \in C_i}(x-\mu_i)^2 $$

Where:

\(C_i\) = cluster
\(\mu_i\) = centroid of that cluster
\(x\) = data point
What does WCSS tell us?

Lower WCSS = points are closer to their centroids = tighter clusters.

For example:

K = 2  → WCSS = 500
K = 3  → WCSS = 250
K = 4  → WCSS = 180
K = 5  → WCSS = 170

As K increases, WCSS always tends to decrease, because having more clusters makes it easier for points to be close to a centroid.

But we don't simply choose the K with the lowest WCSS.

Elbow Method

The Elbow Method uses WCSS to decide a reasonable value of K.

You run K-Means multiple times with different K values:

K = 1
K = 2
K = 3
K = 4
K = 5
K = 6
...

For each K, calculate WCSS.

Then plot:

WCSS
 │
 │ ●
 │   \
 │     ●
 │       \
 │         ●
 │           \
 │             ●──●──●
 │
 └────────────────────── K
    1  2  3  4  5  6

You look for the point where the rate of decrease in WCSS changes significantly.

That's the "elbow."

Why is that point important?

Before the elbow:

Increasing K gives a large improvement in clustering because WCSS decreases significantly.

After the elbow:

Increasing K gives only a small improvement in WCSS.

So we choose the K around the elbow because it gives a good balance between:

fewer clusters + reasonably tight clusters.

Example

Suppose:

K	WCSS
1	1000
2	500
3	250
4	200
5	180
6	170

Here:

K=1 → 1000
K=2 → 500    ↓ 500
K=3 → 250    ↓ 250
K=4 → 200    ↓ 50
K=5 → 180    ↓ 20
K=6 → 170    ↓ 10

The improvement is very large until K=3, then becomes much smaller.

So K = 3 is approximately the elbow.

Interview answer

"WCSS measures the total squared distance between data points and their respective cluster centroids. As the number of clusters increases, WCSS decreases. The Elbow Method calculates WCSS for different values of K and plots K against WCSS. We look for the point where the decrease in WCSS changes from significant to relatively small. This point is called the elbow, and it represents a reasonable trade-off between the number of clusters and cluster compactness."

K-Means++ is an improved way of choosing the initial centroids in K-Means.

The normal K-Means algorithm can randomly choose centroids that are very close to each other. This can lead to poor or slower clustering.

K-Means++ chooses the initial centroids more intelligently and spreads them out.

How K-Means++ works

Suppose K = 3:

Choose the first centroid randomly.
Calculate the distance of every point from the nearest selected centroid.
Give a higher probability to points that are farther away.
Select one of those far-away points as the next centroid.
Repeat until 3 centroids are selected.
Then normal K-Means starts:
Assign points to nearest centroid.
Recalculate centroids.
Repeat until convergence.
