Hierarchical Clustering 

unsupervised techniq that groups similar data points into cluster and create a tree structure 

it use two types 

1.Agglomerative Clustering 
means Bottom-Up

it starts with individual point and move by combining them in one big cluster 

2.Divisive Clustering
Top-Down approach 
it does the opposite starts with all data point in one cluster and seperate them in single cluster


How it Distance Between Clusters Calculated

Linkage Methods:- used to calculate the distance between two clusters


1.Single Linkage/ Min Distance
find the minimum distance

2.complete Linkage/ Max Distance 
find max

3.Avg / Group average 

all points -> average distance

Ward's Method:-
merges two cluster such that the increase in variation within the cluster is as small as possible

Merge the two clusters that make the new cluster most compact

"If I merge these two clusters, how much will the spread/variance increase?"
small variance ->merge
