Imbalance Data

if you are building placement prediction for IIT's student 
then there is higher chance that most of the student with high cgpa and iq will be placed 
there will be very low amount of students which are not placed 

so in that time of problem focus should be more in minority classs why this students are not not placed
it's called Imbalance Data 

your only foucs should not be on accuracy bczo there are more student getting placed so ofcourse accuracy will be 96% that's the problem 

Let's dive into understand

Method 1 :- Undersampling

in this technique majority class (which are getting placed) and minority class so will decrease majority class randomly deleting records from the majority class to match the minority class size

benefit is :- reducing bias faster training

disadv :- important information loss sometime sampling bias

Method 2 :- Oversampling

Increasing minority class samples, to match the majority class size 

how will you fill data :- duplication data

benefit :- reduction bias  focus on minority class

disadv :- Duplication of data may cause overfitting


Method 3 :- SMOTE Synthetic minority over sampling technique 
popular best

- it will increase minority class but not with duplication data instead
