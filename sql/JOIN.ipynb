1.Inner join 
Returns only matching rows from both tables

from customer as c

on c.customer_id = o.customer_id

5.self join
A self join means joining a table with itself 
such as employee -> manager
SELECT 
    e.name AS employee,
    m.name AS manager
FROM employees e
LEFT JOIN employees m
    ON e.manager_id = m.employee_id;

6.CROSS JOIN
returns Cartesian product

Every row of the first table is multipled with every row of the second table

SELECT c.name,p.product_name
from customer c 
CROSS JOIN products p;

### There is no ON condition in cross join

### A CROSS JOIN does not require a foreign key or even a relationship between the tables.
