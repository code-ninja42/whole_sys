## CASE
it is similar to if-else used to implement condtional logic


syntax 

CASE 

    when condition then result 

    when condition then result

    ELse result 
END

SELECT name,
       marks,
       CASE
           WHEN marks >= 80 THEN 'A'
           WHEN marks >= 60 THEN 'B'
           WHEN marks >= 40 THEN 'C'
           ELSE 'Fail'
       END AS grade
FROM student;
