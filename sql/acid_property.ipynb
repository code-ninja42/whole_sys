## ACID Properties
It ensures that db transaction are reliable and consistent

1.Atomicity 
All or Nothing

A transaction is treated as one complete unit 

Example: transferring ₹1,000 from A to B:

A → -₹1,000
B → +₹1,000

Both operations must succeed.

If adding ₹1,000 to B fails, the deduction from A should also be rolled back.

if something fails : ROLLBACK


2.Consistency

Db remains valid before and after the transaction 


3.Isolation

transactions should not interfere with each other independent work 


4.Durability

once commited, the data remains saved

By default
autocommit = 1 enabled
