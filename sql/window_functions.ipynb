## Window Function 
A window function performs a calculation across a set of related rows without combining those row into a single row 

and This is key difference from GROUP BY
with group by multiple rows become group

With a window function, the original rows are kept and the calculation is added as an extra columns

ex :- 
employee | department | salary
---------|------------|-------
Jack     | IT         | 50000
John     | IT         | 70000
Sam      | HR         | 60000
Alex     | HR         | 80000


SELECT
    employee,
    department,
    salary,
    AVG(salary) OVER(PARTITION BY department) AS dept_avg
FROM employees;


employee | department | salary | dept_avg
---------|------------|--------|---------
Jack     | IT         | 50000  | 60000
John     | IT         | 70000  | 60000
Sam      | HR         | 60000  | 70000
Alex     | HR         | 80000  | 70000



syntax 

function() over(
    partition by column
    order by column
)

//These three are important concepts

1.OVER()
it tells SQL that this is a window function

2.PARTITION BY 
Divide rows into groups/windows without mergining them 

3.ORDER BY 
Define the order of rows inside the window


## Main types of window functions

1.Ranking Function

ROw_number():-

Gives a unique sequential number to each row

SELECT
    employee,
    salary,
    ROW_NUMBER() OVER(ORDER BY salary DESC) AS row_num
FROM employees;

employee | salary | row_num
---------|--------|--------
Alex     | 80000  | 1
John     | 70000  | 2
Sam      | 60000  | 3
Jack     | 50000  | 4


RANK() :-

Gives the same rank to tied values but skips ranks

salary | RANK
-------|-----
80000  | 1
80000  | 1
70000  | 3
60000  | 4



DENSE_RANK() :
Same rank for ties but doesn't skip the next rank 


salary | DENSE_RANK
-------|-----------
80000  | 1
80000  | 1
70000  | 2
60000  | 3



2. Aggregate Window Functions

You can use aggregate functions with OVER():

SUM()
AVG()
COUNT()
MIN()
MAX()

3. Value / Navigation Functions

LAG() :- Gets a value from previous row

LEAD() :- Gets a value from next row


SELECT
    employee,
    salary,
    LAG(salary) OVER(ORDER BY salary) AS previous_salary
FROM employees;

SELECT
    employee,
    salary,
    LEAD(salary) OVER(ORDER BY salary) AS next_salary
FROM employees;


FIRST_VALUE() -> First value 

LAST_VALUE() -> Last value
