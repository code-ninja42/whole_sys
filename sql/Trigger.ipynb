## trigger
a trigger is a database object that automatically executes a predefined SQL statement when a specific event occurs on table 

The event can be INSERT,UPDATE,DELETE

Unlike a stored procedure, we don't manually call a trigger. The database automatically executes it when the specified event happens

Before is useful :- Data validation:- check if input should be in between 0 to 100

After Useful :- send alerts, after operation msg

create trigger after_employee_insert
after insert on employees
