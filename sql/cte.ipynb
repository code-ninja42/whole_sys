## CTE - common table expression 
it is a temporary named result during one query execution it do not store inside db make complex query easier to read or perfrom recursive operations 
define using WITH keyword

A CTE is a temporary, named result set that exists only during the execution of a single query. Defined using the WITH keyword, CTEs make complex queries much easier to read and maintain compared to deeply nested subqueries, and they can also be used for recursive logic.

with customer_total AS(
    select customer_id,sum(amount) as total_amount
    from orders 
    group by customer_id
)
select * from customer_total;

A CTE is a temporary named result set that exists only during the execution of a single query, whereas a View is a persistent database object whose query definition remains stored in the database. I use a CTE when I need temporary results or want to simplify a complex query, and I use a View when the query needs to be reused multiple times or when I want to expose a controlled set of rows or columns to users.
