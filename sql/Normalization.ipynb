## normalization 

Normalization is the process to reduce redudance data

1.1NF 
must be atomic 

2.2nf 
must be 1NF 
remove partial depedency
 partial dependency when non-key column depends on only part of composite key 

3.3nf 
must be 2nf
removes transitive dependency 
where non-key column depends on another non-key column

4.4nf
4NF — Fourth Normal Form

4NF (Fourth Normal Form) is used to remove multivalued dependencies.

In simple words:

4NF says that one entity should not have two or more independent multi-valued facts stored in the same table.
