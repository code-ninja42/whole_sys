## SQL 
it is strucutred query lan that used to interact with relational databases
SQL is the standard language to create, read, update, and delete data in relational databases

So, why would a company choose SQL over something like MongoDB?
, SQL databases enforce structure with schemas. So every row in a table follows the same rules. And those acid properties matter for transactions, where correctness is critical.

### 1.DDL - data defination lan
create,alter,drop,Truncate

### 2.DML - data manipulation lan
insert,update,delete

### 3.DQL - data query lan
select

### 4.DCL - data control lan
give permission controls security,access and user permissions

    1.Grant :- assign specific permission to a user account or role

    2.Revoke : cancel or withdraws previously assigned permissions from a specific user or role

### 5.TCL - transaction control lan 
handle transaction integrity and multi-step execution

    1.commit:
    permanently saves all modification made during the current transaction to the database.
    Once committed,the changes cannot be undone

    2.Rollback
   Undoes all modifications made during the current transaction, reverting the database to its last committed state.

   3.savepoint: checkpoint allows you to roll back to that specific point
