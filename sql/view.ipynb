## View 
A view is a virtual table based on the result of a SQL query , it does not normally store the actual data itself instead it stores the query definition 

create view view_name AS
select id,name,department,salary
from employees
where salary > condition;

It is useful security hidding data and for repeatiting task   , complex query

View when you need a permanent, reusable virtual table shared across multiple queries or users, and use a CTE (Common Table Expression) when you need a temporary, readable query block meant for a single statement or recursive logic
