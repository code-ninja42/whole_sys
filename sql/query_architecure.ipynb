## SQL query architecure 

### 1.Parse :
it checks whether the SQL query syntax is valid 

it check thing such as :
* does table exist 
* do the specified columns exist
* are the operations valid

if there is error parser reports and error

### 2.Query Optimizer :
if the query is valid, the query optimizer determines an efficient way to execute it

The optimizer checks:
* index 
* join methods
* join order
* Filtering conditions

### 3.Execution Engine
The execution engine executes the plan selected by the optimizer


### 4.Storage Engine
it stores the actual data 

for ex mysql use Innodb as storage engine
