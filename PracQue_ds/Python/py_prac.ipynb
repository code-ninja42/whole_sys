```python
a = "madam"
if a == a[::-1]:
    print("palindrome")
else:
    print("not")
```

### Find the sum of digits of a number

```python

num = 123 
total = 0
while num > 0:
    digit = num % 10      # get last digit
    total += digit
    
    num = num // 10       # remove last digit
print(total)
```

###  Check whether a number is prime

```python
num = 22

if num > 1:
    for i in range(2, num):
        if num % i == 0:
            print("not prime")
            break
    else:
        print("prime")
else:
    print("not prime")
```

### Armstrong

```python
num = 153
original = num
total = 0

while num > 0:
    digit = num % 10
    total += digit ** 3
    num = num // 10

if original == total:
    print("Armstrong")
else:
    print("Not Armstrong")
```

# Functions

Write a function that accepts any number of numbers and returns their sum.

```python

def sum_number(*args):
    total = 0
    for i in args:
        total += i 
    return total
sum_number(10,20,30,40)
    
```

```python
from functools import reduce 

def sum_number(*args):
    return reduce(lambda a,b:a+b,args)


sum_number(10,20,30,40)
```

```python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance

    def deposit(self, amount):
        self.__balance += amount

    def get_balance(self):
        return self.__balance


account = BankAccount(1000)

account.deposit(500)

print(account.get_balance())
```

```python
from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass
class Dog(Animal):
    def sound(self):
        print("Dog barks")
    print("nice")
d = Dog()
d.sound()
```
